```python
grade_info = pd.DataFrame({
    'grade': ['A', 'B', 'C', 'D', 'F'],
    'grade_description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ]
})

print("\nGrade Information DataFrame:")
print(grade_info)
```

```python
merged = pd.merge(
    df,
    grade_info,
    on='grade',
    how='left'
)

print("\nMerged DataFrame:")
print(merged.head())
```

```python
grade_details = pd.DataFrame({
    'description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ],
    'performance_level': [
        'Very High',
        'High',
        'Medium',
        'Low',
        'Very Low'
    ]
}, index=['A', 'B', 'C', 'D', 'F'])

grade_counts = df['grade'].value_counts().to_frame('student_count')

joined = grade_counts.join(grade_details)

print("\nJoined DataFrame:")
print(joined)
```

```python
df1 = df.iloc[:100]
df2 = df.iloc[100:200]

combined = pd.concat([df1, df2])

print("\nConcatenated DataFrame:")
print(combined.head())

print("\nShape after concatenation:")
print(combined.shape)
```
