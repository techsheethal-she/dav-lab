```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\vbpat\Downloads\archive (2)\Iris.csv")
```

```python
# VIEW THE FIRST 5 ROWS
print("First 5 rows of the dataset:")
print(df.head())
```

```python
#  VIEW THE LAST 5 ROWS
print("\nLast 5 rows of the dataset:")
print(df.tail())
```

```python
# CHECK THE SIZE OF THE DATASET
print("\nShape of the dataset:")
print(df.shape)
```

```python
# VIEW COLUMN NAMES
print("\nColumn names:")
print(df.columns)
```

```python
# CHECK DATASET INFORMATION
print("\nDataset information:")
df.info()
# Check missing values
print(df.isnull().sum())
```

```python
# Remove missing values
df = df.dropna()
```

```python
# Check missing values again
print(df.isnull().sum())
```

```python
# Check duplicate rows
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check duplicates again
print(df.duplicated().sum())
```

```python
# Display statistical summary
print(df.describe())
```

```python
# Set graph style
sns.set_style("whitegrid")
# ---------------- HISTOGRAM ----------------
```

```python
# Create graph area
plt.figure(figsize=(8, 5))
```

```python
# Create histogram
sns.histplot(data=df, x="SepalLengthCm", bins=10)
```

```python
# Add title
plt.title("Distribution of Sepal Length")
```

```python
# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Number of Flowers")
```

```python
# Display graph
plt.show()
# ---------------- BOX PLOT ----------------
```

```python
# Create graph area
plt.figure(figsize=(8, 5))
```

```python
# Create box plot
sns.boxplot(data=df, x="Species", y="SepalLengthCm")
```

```python
# Add title
plt.title("Sepal Length by Species")

# Add X-axis label
plt.xlabel("Species")

# Add Y-axis label
plt.ylabel("Sepal Length (cm)")
```

```python
# Display graph
plt.show()
```

```python
# ---------------- SCATTER PLOT ----------------
# Create graph area
plt.figure(figsize=(8, 5))
```

```python
# Create scatter plot
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)
```

```python
# Add title
plt.title("Sepal Length vs Petal Length")

# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Petal Length (cm)")
```

```python
# Display graph
plt.show()
# ---------------- PAIR PLOT ----------------
```

```python
# Create pair plot
sns.pairplot(df, hue="Species")
```

```python
# Display graph
plt.show()

# ---------------- CORRELATION ----------------
```

```python
# Select numerical columns
numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]
```

```python
# Calculate correlation
correlation = df[numeric_columns].corr()
```

```python
# Display correlation matrix
print(correlation)

plt.figure(figsize=(8, 6))
```

```python
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
